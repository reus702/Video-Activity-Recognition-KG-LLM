```python
from remotezip import RemoteZip
from pathlib import Path

url = (
    "https://researchdata.tuwien.ac.at/api/records/"
    "0ewrv-8cb44/files/data.zip/content"
)

output_dir = Path("../data/raw/reassemble")
output_dir.mkdir(parents=True, exist_ok=True)
print("Cartella di download:", output_dir.resolve())
with RemoteZip(url) as archive:

    files = [
        f for f in archive.namelist()
        if f.endswith(".h5")
    ]

    print(f"File disponibili: {len(files)}")

    # Scarica solamente 10 registrazioni
    for file in files[:5]:

        destination = output_dir / Path(file).name

        if destination.exists():
            continue

        print(f"Downloading: {file}")

        with archive.open(file) as source:
            destination.write_bytes(source.read())
```
